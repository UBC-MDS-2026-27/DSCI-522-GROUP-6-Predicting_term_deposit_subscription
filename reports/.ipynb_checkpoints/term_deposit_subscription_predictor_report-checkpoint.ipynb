## Data

```python
# create the data folders if they don't exist
os.makedirs("../data/raw", exist_ok=True)
os.makedirs("../data/processed", exist_ok=True)

# download the zip file from the UCI Machine Learning Repository
url = "https://archive.ics.uci.edu/static/public/222/bank+marketing.zip"
zip_path = "../data/raw/bank+marketing.zip"

response = requests.get(url)
response.raise_for_status()  # stop with an error if the download failed
with open(zip_path, "wb") as f:
    f.write(response.content)

with zipfile.ZipFile(zip_path) as z:
    z.extractall("../data/raw")

with zipfile.ZipFile("../data/raw/bank.zip") as z:
    z.extractall("../data/raw")
```

```python
# read in the full data set
bank = pd.read_csv("../data/raw/bank-full.csv", sep=";")
```

Before cleaning the data, we checked that the download was complete and that the file has no empty cells or duplicate rows (Table 1). These checks only confirm that the file was read correctly, so they do not use any information that could influence our modelling decisions.

Table 1. Basic checks of the full data set.

```python
pd.DataFrame({
    "Check": ["Number of rows", "Number of columns", "Empty cells", "Duplicate rows"],
    "Result": [
        bank.shape[0],
        bank.shape[1],
        bank.isna().sum().sum(),
        bank.duplicated().sum()
    ]
}).set_index("Check")
```

We removed the call duration variable before modelling. The duration of a call is only known after the call has ended, and by then the outcome (whether the client subscribed) is also known. A model that uses call duration therefore cannot help the bank decide which clients to call, which is the purpose of our analysis. This decision is based on the meaning of the variable, not on the data, so it is safe to make before splitting the data.

```python
# drop duration: it is only known after the call ends,
# so it cannot be used to predict before calling a client
bank_clean = bank.drop(columns=["duration"])
# bank_clean.to_csv("../data/processed/bank_clean.csv", index=False)
```

We then split the data into a training set (80%) and a test set (20%). We split the data before any further exploration, so that the test set does not influence any decision we make. Because only a small share of clients subscribed to a term deposit, we used a stratified split, which keeps the same proportion of subscribers in both sets.

```python
# split the data into training (80%) and test (20%) sets
bank_train, bank_test = train_test_split(
    bank_clean,
    test_size=0.2,
    stratify=bank_clean["y"],
    random_state=522
)

bank_train.to_csv("../data/processed/bank_train.csv", index=False)
bank_test.to_csv("../data/processed/bank_test.csv", index=False)
```

## Data quality in the training set

Although the data set has no empty cells, some categorical variables record missing information as the value "unknown" (Table 2). This affects four variables in the training set: job, education, contact type and the outcome of the previous campaign.

Table 2. Number and percentage of "unknown" values in the training set, for the variables that contain them.

```python
# count "unknown" values in each column of the training set
unknown_counts = (bank_train == "unknown").sum()
unknown_counts = unknown_counts[unknown_counts > 0]

unknown_table = pd.DataFrame({
    "Number of unknown values": unknown_counts,
    "Percent of training rows": (unknown_counts / len(bank_train) * 100).round(1)
}).rename(index={
    "job": "Job",
    "education": "Education",
    "contact": "Contact type",
    "poutcome": "Previous campaign outcome"
})
unknown_table.index.name = "Variable"
unknown_table
```

To understand the "unknown" values in the previous campaign outcome, we compared this variable with the number of days since the client was last contacted in a previous campaign. The data set uses -1 for clients who were not previously contacted (Table 3).

Table 3. Number of clients in the training set with an unknown previous campaign outcome, and how many of them were not previously contacted.

```python
# keep only the clients whose previous campaign outcome is "unknown"
unknown_poutcome = bank_train[bank_train["poutcome"] == "unknown"]

# count how many of them were not previously contacted (pdays == -1)
n_unknown = len(unknown_poutcome)
n_not_contacted = (unknown_poutcome["pdays"] == -1).sum()

pd.DataFrame({
    "Clients with unknown previous outcome": [n_unknown],
    "Of these, not previously contacted": [n_not_contacted],
    "Percent not previously contacted": [round(n_not_contacted / n_unknown * 100, 2)]
}, index=["Training set"])
```

Almost all clients with an unknown previous campaign outcome (29,563 of 29,567, or 99.99%) were not contacted in a previous campaign (Table 3). For this variable, "unknown" therefore does not represent missing information: it means that there was no previous campaign for that client. For the other three variables, the "unknown" values are true missing information, but we kept them as well. Removing these rows would discard a large part of the data, because almost a third of the clients in the training set have an unknown contact type (Table 2). We therefore kept "unknown" as its own category in all four variables, so that the model can learn from it like any other category.

```python
X_train = bank_train.drop(columns="y")
y_train = bank_train["y"]

X_test = bank_test.drop(columns="y")
y_test = bank_test["y"]

def transform_pdays(X):
    X = X.copy()
    X["previously_contacted"] = (X["pdays"] != -1).astype(int)
    return X.drop(columns=["pdays"])

numeric_features = ["age", "balance", "day", "campaign", "previous"]
categorical_features = ["job", "marital", "education", "contact", "poutcome"]
binary_features = ["default", "housing", "loan"]
contacted_features = ["previously_contacted"]
```

```python
numeric_pipe = make_pipeline(StandardScaler())

categorical_pipe = make_pipeline(OneHotEncoder(handle_unknown="ignore"))

binary_pipe = make_pipeline(OrdinalEncoder(categories=[["no", "yes"]] * len(binary_features)))

preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipe, numeric_features),
        ("categorical", categorical_pipe, categorical_features),
        ("binary", binary_pipe, binary_features),
        ("contacted", "passthrough", contacted_features)
    ],
    remainder="drop",
    verbose_feature_names_out=False
)

full_preprocessor = make_pipeline(
    FunctionTransformer(transform_pdays, validate=False),
    preprocessor
)

X_train_processed = full_preprocessor.fit_transform(X_train)
X_test_processed = full_preprocessor.transform(X_test)
```

The Python programming language (Van Rossum and Drake 2009) and the following Python packages were used to perform the analysis: requests (Reitz 2011), zipfile (Van Rossum and Drake 2009), NumPy (Harris et al. 2020), pandas (McKinney 2010), Altair (VanderPlas et al. 2018) and scikit-learn (Pedregosa et al. 2011). The code used to perform the analysis and create this report can be found here: https://github.ubc.ca/mds-2026-27/DSCI_561_lab1_wdz26.git

Firstly, we are going to analyze the features manually with the meaning. The meaning of duration is last contact duration, in seconds (numeric). And the meaning of month is last contact month of year. Therefore, we don't think these two features would have reasonable relationship with the term deposit description. 

Secondly, we will divide the features into different categories. The three main categories are numeric, categorical, and binary. Here are two points need to be mentioned. The first is that binary features are shown in different ways. Features whose values consists of only 0 and 1 are binray, consists of only "no" and "yes" are also binary. The second point is that, pday is a special feature. It has lots of -1 values, and several values are integer. Let's take a look at the description of this feature: number of days that passed by after the client was last contacted from a previous campaign (numeric; -1 means client was not previously contacted). Therefore, we need to specially deal with this feature. Because the meaning of -1 are totally different with the value >= 0, and most of the values are -1, causes an imbalance, so the number of non-negative value would not be that important, I decide to transfer this feature into a new binary feature, which is called previously conducted, where -1 values would be transformed as 0, and the other values would be transformed as 1.



```python
dummy_pipe = make_pipeline(
    full_preprocessor,
    DummyClassifier(strategy="most_frequent")
)

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=123
)

scoring = {
    "accuracy": "accuracy",
    "precision": make_scorer(precision_score, pos_label="yes", zero_division=0),
    "recall": make_scorer(recall_score, pos_label="yes", zero_division=0),
    "f1": make_scorer(f1_score, pos_label="yes", zero_division=0)
}

dummy_scores = cross_validate(
    dummy_pipe,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    return_train_score=True
)

dummy_results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1"],
    "Train": [
        dummy_scores["train_accuracy"].mean(),
        dummy_scores["train_precision"].mean(),
        dummy_scores["train_recall"].mean(),
        dummy_scores["train_f1"].mean()
    ],
    "Validation": [
        dummy_scores["test_accuracy"].mean(),
        dummy_scores["test_precision"].mean(),
        dummy_scores["test_recall"].mean(),
        dummy_scores["test_f1"].mean()
    ]
})
```

```python
reg_pipe = make_pipeline(
    full_preprocessor,
    LogisticRegression(max_iter=2000, random_state=123)
)

reg_scores = cross_validate(
    reg_pipe,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    return_train_score=True
)

reg_results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1"],
    "Train": [
        reg_scores["train_accuracy"].mean(),
        reg_scores["train_precision"].mean(),
        reg_scores["train_recall"].mean(),
        reg_scores["train_f1"].mean()
    ],
    "Validation": [
        reg_scores["test_accuracy"].mean(),
        reg_scores["test_precision"].mean(),
        reg_scores["test_recall"].mean(),
        reg_scores["test_f1"].mean()
    ]
})
```

```python
knn_pipe = make_pipeline(
    full_preprocessor,
    KNeighborsClassifier(n_neighbors=5)
)

knn_scores = cross_validate(
    knn_pipe,
    X_train,
    y_train,
    cv=cv,
    scoring=scoring,
    return_train_score=True
)

knn_results = pd.DataFrame({
    "Metric": ["Accuracy", "Precision", "Recall", "F1"],
    "Train": [
        knn_scores["train_accuracy"].mean(),
        knn_scores["train_precision"].mean(),
        knn_scores["train_recall"].mean(),
        knn_scores["train_f1"].mean()
    ],
    "Validation": [
        knn_scores["test_accuracy"].mean(),
        knn_scores["test_precision"].mean(),
        knn_scores["test_recall"].mean(),
        knn_scores["test_f1"].mean()
    ]
})
```

```python
comparison_df = pd.concat([
    dummy_results.assign(Model="Dummy"),
    reg_results.assign(Model="Logistic Regression"),
    knn_results.assign(Model="KNN")
], ignore_index=True)

display(comparison_df)
```

We decide to take the Dummy Classifier as the baseline. And then we are going to set Logistic Regression and KNearestNeighbour as 2 candidates. All of the three models are trained with the same condition, which is 5-fold cross-validation. 

The DummyClassifier achieved a validation accuracy of 0.883, while the f1 score is 0, which indicates that this model failed to identify customers who subscribed to a term deposit. This is might because of the class imbalance in the dataset. Logistic Regression achieved the highest validation accuracy as 0.894, with the precision as 0.676, with an F1 score of 0.277. KNN achieved a slightly higher F1 score as 0.278, with the recall as 0.192, but its precision was lower as 0.506. In comparison of Logestic Regression and KNearestNeighbour, we can find that the gap of f1 score of train and validation of Logestic Regression, which is about 0.01, is much smaller than the one of K-Nearest Neighbour model, which is about 0.12.

Therefore, we want to choose the Logestic Regression as the best model to use.
